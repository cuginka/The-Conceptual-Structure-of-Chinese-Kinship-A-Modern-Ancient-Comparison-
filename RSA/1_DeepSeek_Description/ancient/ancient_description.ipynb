````python
import pandas as pd
import numpy as np
import os

import time
import json

import openai
#import tiktoken
from dotenv import load_dotenv
load_dotenv()
from openai import OpenAI

api_key = os.getenv("DEEPSEEK_API_KEY")
client = OpenAI(api_key=api_key, base_url="https://api.deepseek.com/v1")
````

````python
def askDeepSeek(messages):
    MODEL = "deepseek-chat"
    response = client.chat.completions.create(
        model= MODEL,
        messages=messages,
        temperature=0,
        stream=False ) # for replication
    return response.choices[0].message.content  # type: ignore


def get_arr(a):
    init = [{
        "role": "user",
        "content": a
    }]
    return init
````

````python
rels_df  = pd.read_excel('./ancient_word.xlsx')
modern_143r = list(rels_df.loc[:,'query'])
modern_143r[0:5]
````

````python
askDeepSeek(get_arr('你好！'))
````

````python
# Iterating through the list of relationships in groups of five.
for i in range(0, 5, 5):
    # Fetching the list of relationships for the current group.
    relationships = modern_143r[i:i+5]

    # Create prompt
    prompt = """请你扮演一个中国文化史专家，对中国古代的人际关系属性进行描述。请注意，一定要在中国古代的语境下讨论这些人际关系，呈现它的内涵和特点。现在我将提供关系列表，请你依次返回对它们关系性质的描述：\n"""
    prompt += '\n'.join(relationships)
    prompt += "\n----\n请返回json格式结果，以关系名为key，关系描述为value，如{‘父母和孩子’:{‘描述’:内容}}。"

    print(prompt)
    
    # Obtaining the results.
    result = askDeepSeek(get_arr(prompt))
    with open(os.path.join('./archieve/example.txt'),'w',encoding='utf-8') as file_group:
         file_group.write(result)
````

````python
# Iterating through the list of relationships in groups of five.
for i in range(0, len(modern_143r), 5):
    # Fetching the list of relationships for the current group.
    relationships = modern_143r[i:i+5]

    # Create prompt
    prompt = """请你扮演一个中国文化史专家，对中国古代的人际关系属性进行描述。请注意，一定要在中国古代的语境下讨论这些人际关系，呈现它的内涵和特点。现在我将提供关系列表，请你依次返回对它们关系性质的描述：\n"""
    prompt += '\n'.join(relationships)
    prompt += "\n----\n请返回json格式结果，以关系名为key，关系描述为value，如{‘父母和孩子’:{‘描述’:内容}}。"

    print(prompt)
    
    # Obtaining the results.
    result = askDeepSeek(get_arr(prompt))
    with open(os.path.join('./archieve/',str(i)+'.txt'),'w',encoding='utf-8') as file_group:
         file_group.write(result)
````

````python
def parse_json_markdown(text):
    # 移除 ```json 和 结尾的 ```，并去除首尾空格
    clean_text = text.replace("```json", "").replace("```", "").strip()
    return json.loads(clean_text)
````

````python
files = os.listdir('./archieve/')
df = pd.DataFrame({'prompt':[],'relation':[],'description':[]})
for i in range(0, len(modern_143r), 5):
    relationships = modern_143r[i:i+5]
    archive_path = './archieve/'+str(i)+'.txt'
    with open(archive_path, 'r', encoding='utf-8') as f:
        content = f.read()
        archieve_results = parse_json_markdown(content)
        temp_df = pd.DataFrame(archieve_results).transpose()
        temp_df.columns = ['description']
        temp_df.insert(0, 'relation', temp_df.index)
        temp_df.insert(0, 'prompt', relationships)
        df = pd.concat([df, temp_df], ignore_index=True)
df.to_csv('./ancient_results.csv',encoding='gb18030')
````
